```python
import heapq

def heuristic(a, b):
  return abs(a[0] - b[0]) + abs(a[1] - b[1])

def astar(grid, start, goal):
  q = [(heuristic(start, goal), 0, start, [start])]
  cost = {start: 0}

  while q:
    f, g, current, path = heapq.heappop(q)

    if current == goal:
      return path, g

    for dx, dy in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
      neighbour = (current[0] + dx, current[1] + dy)

      # Check if the neighbour is within grid boundaries and is walkable (assuming 0 is walkable)
      if 0 <= neighbour[0] < len(grid) and 0 <= neighbour[1] < len(grid[0]):
        if grid[neighbour[0]][neighbour[1]] == 0:
          new_cost = g + 1

          if neighbour not in cost or new_cost < cost[neighbour]:
            cost[neighbour] = new_cost
            f_score = new_cost + heuristic(neighbour, goal)
            heapq.heappush(q, (f_score, new_cost, neighbour, path + [neighbour]))

  return None, float('inf')

# Test case
grid = [
    [0, 0, 0],
    [0, 1, 0],
    [0, 0, 0],
]
start = (0, 0)
goal = (2, 2)

print("Input grid:")
for row in grid:
  print(row)
print("Start:", start)
print("Goal:", goal)

optimal_path, total_cost = astar(grid, start, goal)

if optimal_path:
  print("Optimal path:", optimal_path)
  print("Cost:", total_cost)
else:
  print("No path found.")
```
